```python
import seaborn as sns
df_iris = sns.load_dataset("iris")
print(df_iris.head())
```

```python
df_iris.loc[0:10]
```

```python
df_iris.shape
df_iris.dtypes
```

```python
df_iris.describe()
```

```python
df_iris[(df_iris["petal_length"]>4.5)]
df_iris[(df_iris["species"]=="virginica")]
```

```python
df_iris.groupby("species").agg(
    average_sepalLength=("sepal_length","mean"),
    maximum_petalWidth=("petal_width","max"),
    sd_sepalWidth=("sepal_width","std")
)
```

```python
df_iris.assign(petal_ratio=df_iris["petal_length"]/df_iris["petal_width"])
```

```python
df_titanic = sns.load_dataset("titanic")
df_titanic
```

```python
df_titanic[["sex","age","fare","survived"]]
```

```python
df_titanic[df_titanic["sex"]=="female"]
df_titanic[df_titanic["fare"]>30]
```

```python
df_titanic.groupby("pclass").agg(
    survival_rate=("survived","mean"),
    average_rate=("fare","mean"),
    average_age=("age","mean")
)
```
