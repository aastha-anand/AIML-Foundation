```python
#uniform distribution
import numpy as np
import matplotlib.pyplot as plt

#values from 0 to 10
#parameters lower,upper,no.of random values

values=np.random.uniform(0,10,10_00_000)

plt.hist(values,bins=100,density=True,alpha=0.5)

plt.title("Continous uniform distribtuion graph")
plt.xlabel("x")
plt.ylabel("proabablity density")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

#parameter
mu=70
sigma=10

#x-axis values
x=np.linspace(30,110,1000)

#normal PDF
y=norm.pdf(x,mu,sigma)

#plot the bell curve
plt.plot(x,y,color='black',linewidth=2,label='Normal distribution')

#shade 68%
x1=np.linspace(mu-sigma,mu+sigma,1000)
plt.fill_between(x1,norm.pdf(x1,mu,sigma),color='red',alpha=0.2,label='68%')
#shade 95%
x2=np.linspace(mu-2*sigma,mu+2*sigma,1000)
plt.fill_between(x2,norm.pdf(x2,mu,sigma),color='yellow',alpha=0.2,label='95%')
#shade 99%
x3=np.linspace(mu-3*sigma,mu+3*sigma,1000)
plt.fill_between(x3,norm.pdf(x3,mu,sigma),color='blue',alpha=0.2,label='99%')

plt.title("Normal distribution")
plt.xlabel("Score")
plt.ylabel("probabilty density")
plt.legend()
plt.show()
```
