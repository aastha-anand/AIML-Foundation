```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.set_theme()
```

```python
sns.get_dataset_names()
```

```python
tips=sns.load_dataset("tips")
print(type(tips))
tips.head()
```

```python
#relational plots
#sns.set_theme(style="white")
sns.relplot(
    data=tips,
    x=tips["total_bill"],
    y=tips["tip"],
    hue="smoker",
    size="size",
    style="smoker"
)
```

```python
sns.relplot(
    data=tips,
    x=tips["total_bill"],
    y=tips["tip"],
    kind="line",
)
```

```python
#relational plot
flights=sns.load_dataset("flights")
sns.lineplot(
   data=flights,
    x="year",
    y="passengers"
)
```

```python
#categorial plots - bar plot
tips.head()
sns.barplot(
    data=tips,
    x="day",
    y="tip",
    hue="sex"
)
```

```python
#categorial plots-box plot
tips.head()
sns.boxplot(
    data=tips,
    x="day",
    y="tip"
)
```

```python
#distribution plots
penguins=sns.load_dataset("penguins")
sns.histplot(
    data=penguins,
    x="body_mass_g",
    bins=30
)
```

```python
#matrix plots
flights.head()
flightspivot=flights.pivot(index="month",columns="year",values="passengers")
sns.heatmap(
    flightspivot,
    cmap="coolwarm",
    annot=True,
    fmt="d"
)
```

```python
fig,ax=plt.subplots()

sns.lineplot(
    data=tips,
    x="day",
    y="total_bill",
    hue="sex",
    ax=ax,
    #ci=None
)
ax.set_title("avg bill value for each year")
ax.set_xlabel("days")
ax.set_ylabel("total bill amount")

fig.tight_layout()
```
