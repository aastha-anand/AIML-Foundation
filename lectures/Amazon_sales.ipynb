```python
import json
```

```python
def load_data(filename):
    with open(filename,"r") as f:
        data=json.load(f)
    return data
```

```python
data=load_data("data_store.json")
print(data)
print(type(data))
```

```python
#clean data
def clean_data(data):
    text_to_num={"one":1,
                "two":2,
                "three":3,
                "four":4,
                "five":5}
    cleaned_data=[]
    unique_user=set()
    
    for user in data:
        #standarization of data
        raw_rating=str(user["rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating=text_to_num[raw_rating]

        user["rating"]=raw_rating
      

        age_data=user.get("age")
        if age_data is None:
            user["age"]=None

        name=user["name"].strip()
        if(name in unique_user):
            continue

        unique_user.add(name)
        cleaned_data.append(user)

    return cleaned_data
            
        

    
        
```

```python
cleaned=clean_data(data)
```

```python
print(cleaned)
```

```python
def get_insights(cleaned):
    total_rating=0
    count=0
    user_data={}
    for user in cleaned:
        rating=float(user["rating"])
        total_rating+=rating
        if(rating<3):
            count+=1
        if(rating>=4):
            user_data[user["name"]]="Apple"
        else:
            user_data[user["name"]]="Samsung"
    print (f" Average Rating {total_rating/len(cleaned)}")
    print (f" Poor rating percentage {count/len(cleaned)*100}%")
    print (f"Recommendation{user_data}")

    
    
```

```python
get_insights(cleaned)
```

```python
def recommendation(cleaned):
    recommend=[]
    for user in cleaned:
        current_user={}
        current_user["Name"]=user["name"].strip()
        rating=float(user["rating"])
        
        if(rating>=4):
            current_user["Brand"]="Apple"
        else:
            current_user["Brand"]="Samsung"
        recommend.append(current_user)
    return recommend
        
```

```python
recommendation(cleaned)
```
