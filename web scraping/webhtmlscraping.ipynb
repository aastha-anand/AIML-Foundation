```python
if res.status_code==200:
    #text,
    print(res.headers)
with open("scraped_data/data1.html","w",encoding="utf-8") as f:
    f.write(res.text)
```

```python
with open("scraped_data/data1.html","r") as f:
    html_content=f.read()

soup=BeautifulSoup(html_content,"html.parser")
```

```python
all_h3=soup.find_all("h3")
for h3 in all_h3:
    names=h3.get_text(strip=True)
    print(names)
```

```python
countries_info=[]
for h3 in all_h3:
    names=h3.get_text(strip=True)
    population=(h3.find_next("div").select("span.country-population")[0].get_text(strip=True))
    countries_info.append([names,population])
```

```python
countries_info
```

```python
import pandas as pd
df=pd.DataFrame(countries_info,columns=["Name","Population"])
```
